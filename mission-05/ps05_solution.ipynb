## Part 1: Systems Architect — HNSW Vector Proximity Index (Hassaan Ahmad Khawaja)
Hierarchical Navigable Small World skip-graph construction and logarithmic beam traversal.

```python
from importlib import import_module
import numpy as np
hnsw_mod = import_module("01_HNSW_Vector_Index")
WinogradHNSWIndex = hnsw_mod.WinogradHNSWIndex

np.random.seed(999)
test_dim = 64
idx = WinogradHNSWIndex(vector_dim=test_dim, max_edges=8, ef_build=32)
for i in range(120):
    idx.insert(i, np.random.randn(test_dim).astype(np.float32))

q = np.random.randn(test_dim).astype(np.float32)
matches = idx.query(q, top_k=3)
print("=== SQUADRON WINOGRAD: HNSW VECTOR INDEX VERIFICATION ===")
print("Top Astrogation Proximity Results:")
for nid, score in matches:
    print(f"  Target ID: {nid:03d} | Cosine Proximity: {score:.4f}")
print("HNSW Skip-Graph Verified!")
```

## Part 2: Communications & Protocol Officer — BM25 & Reciprocal Rank Fusion (Muhammad Usman Maratib)
Exact serial preservation (PR-882-HEX) and dual-path rank fusion (k=60).

```python
bm25_mod = import_module("02_BM25_and_RRF_Combiner")
HybridRetrievalPipeline = bm25_mod.HybridRetrievalPipeline

sample_corpus = [
    "AURA Subsystem 05 associative memory core undergoing Crab Pulsar calibration.",
    "Primary magnetic deflector shield harmonic code is MAG-SHIELD-77B at 30.22 Hz.",
    "Warp vector plasma manifold alignment requires exact code PR-882-HEX.",
    "Cryo-coolant emergency valve designated VLV-904-CRY for antimatter chill loop.",
    "General telemetry archives indicate no coronal flares detected in Taurus filament."
]
pipeline = HybridRetrievalPipeline(sample_corpus, rrf_k=60)
simulated_dense = [0, 2, 1, 3, 4]
fused = pipeline.retrieve("PR-882-HEX plasma alignment", simulated_dense, top_results=2)
print("=== SQUADRON WINOGRAD: BM25 & RECIPROCAL RANK FUSION ===")
for item in fused:
    print(f"Doc {item['doc_id']} | RRF Score: {item['rrf_score']:.6f} | BM25: {item['bm25_score']:.3f}")
    print(f"  Excerpt: {item['document_text']}")
print("Hybrid Retrieval Pipeline Verified!")
```
