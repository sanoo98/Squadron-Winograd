## Part 1: Systems Architect — Tree-of-Thoughts (ToT) Beam & Backtracking Solver
**Lead Station Cadet:** Hassaan Ahmad Khawaja (Systems Architect)  
Implements Breadth-First Beam Search (ToT-BFS) and Depth-First Backtracking Search (ToT-DFS) with hard compute expansion budget enforcement ($N_{exp} \le 20$).

```python
root = ThoughtNode(thought_state="Initial Nav Vector: Ingress at Sgr A* r=10 r_s")

def generator(node):
    return [
        f"{node.thought_state} -> Retro-burn thrusters",
        f"{node.thought_state} -> Direct radial descent (Deadly)",
        f"{node.thought_state} -> Relativistic frame-drag slingshot (Optimal)"
    ]

def evaluator(thought, parent):
    if "Direct radial" in thought:
        return 0.15
    elif "slingshot" in thought:
        return 0.94
    return 0.60

solver = TreeOfThoughtsSolver(beam_width=2, max_expansions=10, prune_threshold=0.40)
best_bfs = solver.bfs_beam_search(root, generator, evaluator, max_depth=3)

print("=== Winograd ToT BFS Beam Search Execution ===")
print(f"Top Thought Trajectory End: {best_bfs.thought_state}")
print(f"Final Step Score: {best_bfs.step_score:.2f}")
print(f"Expansions Count: {solver.expansion_count} (Limit: {solver.max_expansions})")
print(f"Pruned Subtrees Count: {solver.pruned_branches_count}")
```

## Part 2: Communications & Protocol Officer — 4-Phase MCTS & PUCT Engine
**Lead Station Cadet:** Muhammad Usman Maratib (Communications & Protocol Officer)  
Executes Selection (PUCT), Expansion, Simulation (Rollout), and Backpropagation over LLM policy priors.

```python
root = MCTSNode(state="Sgr A* Horizon Ingress")
engine = MCTSProtocolEngine(c_puct=1.414)

def mock_cand(node):
    return [
        (f"{node.state} -> Route 1: Magnetic reconnection channel", 0.7),
        (f"{node.state} -> Route 2: High shear accretion path", 0.1),
        (f"{node.state} -> Route 3: Standard orbital clearance", 0.2)
    ]

def mock_eval(node):
    if "Magnetic reconnection" in node.state:
        return 0.92
    elif "High shear" in node.state:
        return 0.10
    return 0.50

best_node = engine.run_search(root, iterations=30, candidate_generator=mock_cand, rollout_evaluator=mock_eval)

print("=== Winograd MCTS 4-Phase Protocol Search ===")
print(f"Root Visits: {root.visit_count}")
print(f"Optimal Action Selected: {best_node.state.split('-> ')[-1]}")
print(f"Visits Allocated: {best_node.visit_count}, Mean Action Value Q: {best_node.q_value:.4f}")
```
