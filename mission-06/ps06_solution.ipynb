```python
logger = MarkovDecisionTrajectoryLogger(gamma=0.95)

# Standard return verification
test_rewards = [10.0, 0.0, -5.0, 20.0]
g0 = MarkovDecisionTrajectoryLogger.calculate_discounted_return_from_rewards(test_rewards, gamma=0.95)
print("=== MDP Return Calculation ===")
print(f"Rewards Sequence: {test_rewards} with gamma=0.95")
print(f"Discounted Return G_0: {g0:.4f} (Exact: 22.6350)")

# Flight log transitions
logger.record_transition("STANDBY", "Init", 10.0, "SENSING")
logger.record_transition("SENSING", "Probe", 0.0, "REASONING")
logger.record_transition("REASONING", "Tune", -5.0, "DISPATCHING")
logger.record_transition("DISPATCHING", "Verify", 20.0, "EVALUATING")
logger.record_transition("EVALUATING", "Loop", 5.0, "REASONING")

matrix, states = logger.estimate_transition_matrix()
print(f"\nIdentified States: {states}")
print(f"Empirical Transition Probability Matrix Shape: {matrix.shape}")
```
